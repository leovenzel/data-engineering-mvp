```python
# ==============================================================================
# EXPLORAÇÃO LEITURA-APENAS DA BASE BRUTA (INSIDE AIRBNB RJ)
# ==============================================================================

# 1. Caminho do arquivo bruto salvo no Volume do Unity Catalog
raw_path = "/Volumes/workspace/default/raw_data/listings.csv.gz"

# 2. Leitura dos dados brutos com suporte a multilinha
df_raw = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("multiline", "true") \
    .option("quote", '"') \
    .option("escape", '"') \
    .load(raw_path)

# ------------------------------------------------------------------------------
# A. VOLUMETRIA (CONTAGEM DE LINHAS E COLUNAS)
# ------------------------------------------------------------------------------
total_linhas = df_raw.count()
total_colunas = len(df_raw.columns)

print(f"📊 VOLUMETRIA DA BASE BRUTA:")
print(f" - Total de Linhas (Anúncios): {total_linhas:,}")
print(f" - Total de Colunas: {total_colunas}")
print("-" * 50)

# ------------------------------------------------------------------------------
# B. LISTAGEM COMPLETA DAS 75 COLUNAS BRUTAS E SEUS TIPOS
# ------------------------------------------------------------------------------
print("📋 RELAÇÃO DE TODAS AS COLUNAS E TIPOS INFERIDOS:")
for idx, (col_name, dtype) in enumerate(df_raw.dtypes, start=1):
    print(f"{idx:02d}. {col_name:<35} | Tipo: {dtype}")

# ------------------------------------------------------------------------------
# C. VISUALIZAÇÃO INTERATIVA DAS PRIMEIRAS LINHAS
# ------------------------------------------------------------------------------
# Exibe a tabela na interface gráfica do Databricks para rolar horizontal e verticalmente
display(df_raw)
```
